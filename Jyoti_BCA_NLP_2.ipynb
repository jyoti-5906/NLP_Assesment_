```python
## Removing Punctuation
import string
text = "ARYANSHI is$$$$$ a@ good frnd of mine"
Jyoti= word_tokenize(text)
print(Jyoti)

words= []
for i in Jyoti:
    if i not in string.punctuation:
        words.append(i)
print(words)

```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')

text = "Code is like humor: when you have to explain it, it’s bad."

J = word_tokenize(text)

words = []

for i in J:
    if i.lower() not in stopwords.words('english'):
        words.append(i)

print(words)
```

```python
# Spelling checking 1
!pip install pyspellchecker
from spellchecker import SpellChecker
spell = SpellChecker()
text = "i huv a gud, bd, and ugly situation"
jk = text.split()
correct = []
for i in jk:
    correct.append(spell.correction(i))
print(correct)
```
