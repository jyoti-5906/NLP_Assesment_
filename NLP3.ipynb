```python
from textblob import TextBlob

text = "I lov programming. It is amazng and fun."
blob = TextBlob(text)

print("Original Text:")
print(text)

print("\nCorrected Text:")
print(blob.correct())
```

Word-by-word spelling check

```python
from textblob import Word

words= ["lov", "amazng", "fun", "programing", "pthon", "dta", "scienc"]

for word in words:
    corrected_word = Word(word).correct()
    print(word, "->", corrected_word)
```

STEMMING


```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ['plays', 'playing', 'playful', 'players',
         'written', 'happiness', 'child', 'children']

for i in words:
    print(i, "-->", ps.stem(i))
```

Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lemm = WordNetLemmatizer()
words = ['running', 'ate', 'better', 'fishing', 'flies']
for i in words:
    print(i, "-->", lemm.lemmatize(i))
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemm = WordNetLemmatizer()

text = """I am running in the park. He ate an apple.
She is better than me. They are fishing in the lake.
The flies are annoying."""

Jyoti = word_tokenize(text)

for i in Jyoti:
    print(i, "-->", lemm.lemmatize(i, pos='v'))
```

UNICODE NORMALIZATION

```python
import unicodedata
text = "José María André ① ② ③"
text = unicodedata.normalize('NFKC', text)
print(text)  
```
